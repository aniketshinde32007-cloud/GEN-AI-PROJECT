# Phase 3 (prepare for chunking)

What is chunking?

Chunking means splitting a large document into smaller pieces so the AI can search for relevant financial information instead of processing the entire document for every question.

For example:

Chunk 1: Income statement

Chunk 2: Balance sheet

Chunk 3: Cash flow statement

Later, we can split these into smaller, overlapping text sections to improve retrieval.

```python

# Step 4: Split the financial knowledge base into chunks

chunk_size = 2000                                          #Each chunk can contain up to 2,000 characters.
chunk_overlap = 200                                        #Consecutive chunks share 200 characters.

chunks = []                                                #Creates an empty list to store the chunks.

start = 0                                                  #Starts reading from the first character.

while start < len(financial_knowledge_base):                #Repeats while there is text left to process.

    end = start + chunk_size                                #Calculates where the current chunk should end.

    chunk = financial_knowledge_base[start:end]             #Extracts a section of text.

    chunks.append(chunk)                                    #Adds that section to the list.

    start = end - chunk_overlap                             #Moves forward while keeping 200 characters from the previous chunk.

print("Chunking completed successfully!")
print("Total characters:", len(financial_knowledge_base))
print("Chunk size:", chunk_size)
print("Chunk overlap:", chunk_overlap)
print("Total chunks created:", len(chunks))

print("\n===== FIRST CHUNK PREVIEW =====")
print(chunks[0])
```

2. Next step: Create embeddings

What is an embedding?

An embedding converts text into a list of numbers that represents its meaning. This allows a system to find text related to a question, even when the exact words differ.

For example:

Financial chunk: “Net Income and profitability”

User question: “How much profit did Reliance make?”

Embeddings help the system identify that these may be related.

```python

# Step 5.2: Load the embedding model

from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2") # "all-MiniLM-L6-v2"- Loads a pretrained model that converts text into numerical vectors.

print("Embedding model loaded successfully!")
```

```python

# Convert all text chunks into numerical embeddings

embeddings = embedding_model.encode(chunks)           #embedding_model.encode(chunks) -Converts every chunk into a numerical vector.

print("Embeddings created successfully!")
print("Number of chunks:", len(chunks))               #Counts the number of text chunks.
print("Embedding shape:", embeddings.shape)           #Shows the number of vectors and numbers per vector.
print("First embedding preview:", embeddings[0][:10]) #Displays the first 10 numbers from the first vector.
```

#Next: Phase 3 — Step 6: Create the FAISS vector database

What is FAISS?

FAISS is a library that searches numerical vectors efficiently. We'll use it to find the financial chunks most relevant to a user's question.

For example, if you ask, “What is Reliance's debt?”, FAISS will help locate chunks that contain relevant debt information.

```python

# Step 6.1: Install the FAISS library

!pip -q install faiss-cpu                #-q reduces installation messages.

print("FAISS installed successfully!")
```

```python

# Step 6.2: Create a FAISS vector index

import faiss
import numpy as np

# Convert embeddings into the format FAISS expects
embeddings = np.array(embeddings).astype("float32")

# Get the number of dimensions in each embedding
embedding_dimension = embeddings.shape[1]


# Create a FAISS index for similarity search
faiss_index = faiss.IndexFlatL2(embedding_dimension)
# Store the embeddings in the index
faiss_index.add(embeddings)

print("FAISS index created successfully!")
print("Embedding dimensions:", embedding_dimension)
print("Vectors stored in FAISS:", faiss_index.ntotal)
```

```python

# Step 6.3: Search the financial knowledge base

question = "What is Reliance's net income?"

# Convert the question into an embedding
question_embedding = embedding_model.encode([question]) #embedding_model.encode([question]) converts it into a 384-dimensional vector.

# FAISS expects float32 vectors
question_embedding = np.array(question_embedding).astype("float32")   #question_embedding - The numerical vector representing your question.
                                                                      #np.array(...)      -  Makes sure the data is stored as a NumPy array.
                                                                      #.astype("float32") -  Converts the numbers to 32-bit floating-point values.

# Retrieve the 3 closest chunks
distances, indices = faiss_index.search(question_embedding, k=3) #faiss_index.search(..., k=3) searches for the three closest vectors.
                                                                 #distances tells us how far the vectors are from the question vector; for this L2 index, lower values mean closer vectors.
                                                                 #indices tells us which chunks were found.
print("Question:", question)
print("\n===== RETRIEVED FINANCIAL CHUNKS =====")

for rank, index in enumerate(indices[0]):
    print(f"\nResult {rank + 1}")                                #rank + 1 converts 0, 1, 2 into 1, 2, 3.
    print("Distance:", distances[0][rank])
    print(chunks[index][:700])                                   #Display the first 700 characters of the retrieved chunk.
    print("-" * 50)

#Main Point to remember Smaller distance = closer match.
```

```python

# Step 6.4: Find the net-income rows directly

# Find financial metric names containing "Net Income"
net_income_rows = income_statement[                      #income_statement — the financial data we collected from Yahoo Finance.
    income_statement.index.str.contains(                 #.index.str.contains(...) — searches the metric names for “Net Income”.
        "Net Income", case=False, na=False               #case=False — matches uppercase or lowercase text.
    )                                                    #na=False — safely ignores missing metric names.
]

# Display the matching rows
display(net_income_rows.T)                               #net_income_rows — stores all matching rows.
                                                         #.T — transposes the table, making reporting dates easier to read vertically.
```

```python

# Step 6.5: Create metric-based financial chunks

financial_chunks = []

statements = [
    ("Income Statement", income_statement),
    ("Balance Sheet", balance_sheet),
    ("Cash Flow Statement", cash_flow)
]

for statement_name, statement in statements:
    for metric, row in statement.iterrows():

        # Keep only values that are available
        available_values = row.dropna()

        if available_values.empty:
            continue

        # Keep each reporting date beside its value
        values_text = "; ".join(
            f"{period}: {value}"
            for period, value in available_values.items()
        )

        chunk = (
            f"Company: {company_name}\n"
            f"Ticker: {ticker_symbol}\n"
            f"Statement: {statement_name}\n"
            f"Metric: {metric}\n"
            f"Values: {values_text}"
        )

        financial_chunks.append(chunk)

print("Metric-based chunks created:", len(financial_chunks))
print("\nExample chunk:\n")
print(financial_chunks[0])
```

```python

# Create embeddings for the improved financial chunks

financial_embeddings = embedding_model.encode(financial_chunks)

print("Embeddings created successfully!")
print("Number of chunks:", len(financial_chunks))
print("Embedding shape:", financial_embeddings.shape)



#Why are we doing this? The old FAISS index contains embeddings for the original 13 character-based chunks.
#We need new embeddings for the 174 metric-based chunks so the improved index can search the new format.
```

```python

# Step 6.7: Build a new FAISS index

#We need to replace the old index, which contained the original 13 chunks, with a new index containing your 174 improved chunks.

import faiss
import numpy as np

# Convert embeddings to the format FAISS expects
financial_embeddings = np.array(financial_embeddings).astype("float32")  #“I converted the embeddings to 32-bit floating-point numbers because FAISS expects this data format for the vector index.”

# Get the number of dimensions in each embedding
embedding_dimension = financial_embeddings.shape[1]

# Create a fresh FAISS index
financial_faiss_index = faiss.IndexFlatL2(embedding_dimension)

# Store all 174 embeddings in the index
financial_faiss_index.add(financial_embeddings)

print("New FAISS index created successfully!")
print("Embedding dimensions:", embedding_dimension)
print("Vectors stored:", financial_faiss_index.ntotal)
```

```python

# Step 6.8: Search the improved financial index

question = "What is Reliance's net income?"

# Convert the question into an embedding
question_embedding = embedding_model.encode([question])

# Convert the embedding to FAISS-compatible format
question_embedding = np.array(question_embedding).astype("float32")

# Search for the 3 closest financial chunks
distances, indices = financial_faiss_index.search(
    question_embedding, k=3
)

print("Question:", question)
print("\n===== RETRIEVED FINANCIAL METRICS =====")

for rank, index in enumerate(indices[0]):
    print(f"\nResult {rank + 1}")
    print("Distance:", distances[0][rank])
    print(financial_chunks[index])
    print("-" * 50)
```

1. What is an Embedding?

Simple meaning: An embedding converts text into numbers so a computer can understand and compare its meaning.

2. What is Float?

Simple meaning: A float is a number that can contain a decimal point.

3. What is a Separator?

Simple meaning: A separator is something used to divide different pieces of information and make them easier to read.

4. What is Distance?

Simple meaning: Distance tells us how close two sets of embedding numbers are to each other.

For the L2 distance used in your project:

Lower distance = the records are closer in numerical space.

Higher distance = the records are farther apart.


```python

import numpy as np

def ask_financial_question(question, top_k=3):

    # Step 1: Convert question into an embedding
    question_embedding = embedding_model.encode([question])
    question_embedding = np.array(question_embedding).astype("float32")

    # Step 2: Retrieve relevant financial records
    distances, indices = financial_faiss_index.search(
        question_embedding, k=top_k
    )

    retrieved_chunks = [
        financial_chunks[i]
        for i in indices[0]
        if i >= 0
    ]

    context = "\n\n".join(retrieved_chunks)

    # Step 3: Prepare the prompt
    prompt = f"""
You are a financial data assistant.
Answer using only the financial data below.
Do not invent missing values.
Mention the metric name and reporting date.
If the exact metric is unavailable, say so.
Do not give investment advice.

Financial data:
{context}

Question: {question}
"""

    # Step 4: Call Gemini and handle errors
    try:
        response = client.models.generate_content(
            model="gemini-3.8-flash",
            contents=prompt
        )
        return response.text, retrieved_chunks

    except Exception as error:
        if "429" in str(error) or "RESOURCE_EXHAUSTED" in str(error):
            print("Gemini daily quota exceeded. Try again after it resets.")
        else:
            print("Gemini API error:", error)

        return None, retrieved_chunks
```

```python
#Step 11.6 — Improve the number formatting
#The number is correct based on the conversion you performed, but let's fix one important display detail: Indian-style commas should show ₹9,64,693.00 crore.

def format_inr_crore(value):
    crore = value / 10_000_000
    formatted = f"{crore:,.2f}"

    # Convert comma placement to Indian numbering style
    whole, decimal = formatted.split(".")
    last_three = whole[-3:]
    remaining = whole[:-3]

    if remaining:
        remaining = remaining.replace(",", "")
        parts = []
        while len(remaining) > 2:
            parts.insert(0, remaining[-2:])
            remaining = remaining[:-2]
        if remaining:
            parts.insert(0, remaining)
        whole = ",".join(parts + [last_three])
    else:
        whole = last_three

    return f"₹{whole}.{decimal} crore"

print(format_inr_crore(9646930000000))
```
